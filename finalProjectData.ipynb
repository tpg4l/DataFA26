## Dataset description:
playerId - string: The unique id given to the player that can be used to identify them even after changing their screen name.
parcelCount - int: The amount of parcels or land plots each player has.
lastPurchaseAt - timestamp: This is the last time a player has purchased a parcel **in the city of Kent**, it is kept in UDT or 5 hours ahead of EST.
rank - int: Leaderboard position of the player in Kent. Whoever has the most plots is the current "mayor" of Kent.
